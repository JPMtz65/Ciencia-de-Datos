```r
#install.packages("dplyr")
library(dplyr)

df <- data.frame(
  cliente = c("A", "B", "C"),
  ventas = c(1000, 2500, 1800)
)

df %>%
  arrange(desc(ventas))
```

```r
library(ggplot2)

ggplot(df, aes(x = cliente, y = ventas)) +
  geom_col() +
  theme_minimal()
```
