# Clases y objetos

En este apunte vamos a definir los conceptos de clase, clase abstracta, herencia, encapsulamiento y polimorfismo.

**Definición**: Un objeto es una instancia de una clase.

```python
import numpy as np 
import numbers
from abc import ABC, abstractmethod
import math
```

## Clases

```python
#Todas las clases se definen con la palabra reservada class, seguida del nombre de la clase. 
class Vector:
    
    #Metodo constructor, se define con self y los argumentos que tomara la clase de manera general. 
    def __init__(self, v: list):
        
        #Validar cada entrada sea un numero
        if not all(isinstance(i, numbers.Number) for i in v):
            raise ValueError("Todos los elementos del vector deben ser numeros.")
        
        self.v = v 
    
    #Metodo de multiplicar por escalares
    def mult_escalar(self, a):
        #Validar a sea un numero
        if not isinstance(a, numbers.Number):
            raise TypeError("El escalar a multiplicar debe ser un numero.")
        self.v = [a*i for i in self.v]
        return self.v
        
    #Metodo para sumar un vector
    def sum_vector(self, u: list): 
        
        #Validar tipo de entrada
        if not isinstance(u, list):
            raise TypeError("El parametro 'u' debe ser una lista.")
            
        #Validar que las dimensiones coincidan para poder sumarlos
        if len(self.v) != len(u):
            raise ValueError(f"Las dimensiones no coinciden.")
        self.v = [i + j for i,j in zip(self.v, u)]
        return self.v    
    
    #Metodo para obtener la norma de un vector
    def norma(self, p):
        
        #Manejo del caso de norma infinita
        if p == "infinity" or p == np.inf:
            return np.max([np.abs(i) for i in self.v])
            
        #Validar que p sea un numero valido para la norma
        if isinstance(p, numbers.Number):
            if p < 1:
                raise ValueError("El valor de p para la norma debe ser mayor o igual a 1.")
            return sum([np.abs(i)**p for i in self.v])**(1/p)
            
        #Si no es numero ni infinito, lanzar error
        raise ValueError("El parametro 'p' debe ser un número >= 1 o la cadena 'infinity'.")
        
    #Metodo para vector unitario
    def vector_unitario(self, p):
        
        #Reutilizamos el método norma para no duplicar código
        norm = self.norma(p)
        
        # Prevenir división por cero (si es el vector nulo)
        if norm == 0:
            raise ZeroDivisionError("No se puede obtener el vector unitario de un vector nulo (norma 0).")
            
        u = [i / norm for i in self.v]
        return u
        
```

## Clases abstractas

```python
#Clase abstracta
class Figura(ABC):
    
    #Ponemos los decoradores a cada metodo ya que son metodos abstractos
    @abstractmethod
    def area(self) -> float:
        pass
    
    @abstractmethod
    def perimetro(self) -> float:
        pass
```

```python
#Clase hija Rectangulo
class Rectangulo(Figura):
    def __init__(self, base: float, altura: float):
        self.base = base
        self.altura = altura
        
    #Metodo heredado
    def area(self) -> float:
        return self.base * self.altura
    
    #Metodo heredado    
    def perimetro(self) -> float:
        return 2 * (self.base + self.altura)
        
    #Metodo propio: verifica si el rectangulo es un cuadrado perfecto
    def es_cuadrado(self) -> bool:
        return self.base == self.altura
```

```python
#Clase hija Triangulo
class Triangulo(Figura):
    def __init__(self, lado_a: float, lado_b: float, lado_c: float):
        
        #Validacion matemática: Desigualdad triangular
        if lado_a + lado_b <= lado_c or lado_a + lado_c <= lado_b or lado_b + lado_c <= lado_a:
            raise ValueError("Los lados proporcionados no forman un triangulo valido.")
        
        self.a = lado_a
        self.b = lado_b
        self.c = lado_c
    
    #Metodo heredado    
    def area(self) -> float:
        #Formula de Heron
        s = self.perimetro() / 2
        return math.sqrt(s * (s - self.a) * (s - self.b) * (s - self.c))
    
    #Metodo heredado    
    def perimetro(self) -> float:
        return self.a + self.b + self.c
        
    #Metodo propio: Clasifica el triangulo segun sus lados
    def tipo_triangulo(self) -> str:
        if self.a == self.b == self.c:
            return "Equilatero"
        elif self.a == self.b or self.b == self.c or self.a == self.c:
            return "Isosceles"
        return "Escaleno"
```

```python
#Clase hija Circulo
class Circulo(Figura):
    def __init__(self, radio: float):
        if radio <= 0:
            raise ValueError("El radio debe ser mayor a 0.")
        self.radio = radio
        
    def area(self) -> float:
        return math.pi * (self.radio ** 2)
        
    def perimetro(self) -> float:
        return 2 * math.pi * self.radio
        
    #Metodo propio Obtener diámetro
    def diametro(self) -> float:
        return self.radio * 2
        
    #Metodo propio Obtener la curvatura 
    def curvatura(self) -> float:
        return 1 / self.radio
```

```python
#Clase hija Rombo 
class Rombo(Figura):
    def __init__(self, diagonal_mayor: float, diagonal_menor: float):
        
        self.D = diagonal_mayor
        self.d = diagonal_menor
        
        #Guardamos el lado para el perimetro
        self.lado = math.sqrt((self.D / 2)**2 + (self.d / 2)**2)
        
    def area(self) -> float:
        return (self.D * self.d) / 2
        
    def perimetro(self) -> float:
        return 4 * self.lado
    
    #Metodo propio Obtener lado
    def obtener_lado(self):
        return self.lado
```
